Définition du critère principal

Un patient est répondeur s’il satisfait les deux conditions :

- Rétention
→ au moins une observation avec VISITNUM ≥ 13

- Abstinence opioïdes
→ tous les tests opioïdes disponibles à VISITNUM ≥ 13 sont NEGATIVE
(OPIATE, MORPHINE, METHADON)

Critère d’évaluation principal de l’efficacité

Le critère d’évaluation principal de l’efficacité repose sur un score composite, combinant :

Le maintien dans le traitement jusqu’à la fin de la période de désintoxication de 13 jours ;

L’abstinence aux opiacés, définie par un dernier échantillon d’urine négatif aux opiacés à l’issue de cette période.

Un patient répondeur est ainsi défini comme tout participant ayant :

complété l’intégralité de la phase de désintoxication de 13 jours, et

fourni un dernier échantillon d’urine négatif aux opiacés.

Ce score composite vise à refléter simultanément :

la tolérance et l’acceptabilité du traitement (via le maintien),

et l’objectif clinique d’abstinence à court terme.

Cette définition présente l’avantage méthodologique de ne poser aucun problème de gestion des données manquantes :
tout patient ne se présentant pas à la fin de la période de traitement ou ne fournissant pas d’échantillon d’urine est, par définition, considéré comme non-répondeur.

Hypothèses statistiques

Hypothèse nulle (H₀) :
La proportion de patients répondeurs est identique dans le groupe traité par buprénorphine/naloxone (BUP/NX) et dans le groupe traité par clonidine.

Hypothèse alternative (H₁) :
La proportion de patients répondeurs est significativement plus élevée dans le groupe BUP/NX que dans le groupe clonidine.

L’analyse statistique vise donc à tester la supériorité du traitement par BUP/NX par rapport à la clonidine sur ce critère composite.
